```python
import sys
import inspect

call_stack = []

def trace_calls(frame, event, arg):
    if event == "call":
        func_name = frame.f_code.co_name
        filename = frame.f_code.co_filename

        # filter only your project files
        if "src" in filename or "main.py" in filename:
            indent = "  " * len(call_stack)
            print(f"{indent}→ {func_name}")

            call_stack.append(func_name)

    elif event == "return":
        if call_stack:
            call_stack.pop()

    return trace_calls


def run_with_trace(func):
    sys.settrace(trace_calls)
    try:
        func()
    finally:
        sys.settrace(None)
```

```python
from main import main

run_with_trace(main)
```
