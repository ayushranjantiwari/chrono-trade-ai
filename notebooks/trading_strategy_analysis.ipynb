```python
def evaluate_strategy(df):

    returns = df["Strategy_Return"]

    total_return = df["Portfolio"].iloc[-1] / df["Portfolio"].iloc[0] - 1
    volatility = returns.std()
    sharpe = returns.mean() / (volatility + 1e-8)

    # drawdown
    cum = df["Portfolio"]
    peak = cum.cummax()
    drawdown = (cum - peak) / peak
    max_dd = drawdown.min()

    return {
        "Total_Return": total_return,
        "Sharpe": sharpe,
        "Max_Drawdown": max_dd
    }

print("SARIMAX:", evaluate_strategy(sarimax_strategy))
print("XGBoost:", evaluate_strategy(xgb_strategy))
```

```python
comparison = pd.DataFrame([
    {"Model": "SARIMAX", **evaluate_strategy(sarimax_strategy)},
    {"Model": "XGBoost", **evaluate_strategy(xgb_strategy)}
])

comparison
```
